```python
import torch
import time

def benchmark_dtype(dtype, size=4096, iters=100):
    a = torch.randn(size, size, dtype=dtype, device="cuda")
    b = torch.randn(size, size, dtype=dtype, device="cuda")
    torch.cuda.synchronize()
    t0 = time.time()
    for _ in range(iters):
        c = a @ b
    torch.cuda.synchronize()
    return (time.time() - t0) / iters

print(f"fp16: {benchmark_dtype(torch.float16)*1000:.2f} ms")
print(f"bf16: {benchmark_dtype(torch.bfloat16)*1000:.2f} ms")
print(f"fp32: {benchmark_dtype(torch.float32)*1000:.2f} ms")
```
